```python
from google.colab import drive
import zipfile
import io
import pandas as pd

drive.mount('/content/drive')

zip_path = "/content/drive/MyDrive/OLIVES/olive.zip"

print("="*80)
print("EXTRACTING AND ANALYZING CLINICAL INFORMATION")
print("="*80)

# Extract the labels zip to examine clinical data
with zipfile.ZipFile(zip_path, 'r') as outer_zip:
    print("\n⏳ Reading OLIVES_Dataset_Labels.zip...")
    labels_zip_data = outer_zip.read('OLIVES_Dataset_Labels.zip')

    with zipfile.ZipFile(io.BytesIO(labels_zip_data), 'r') as labels_zip:
        # List all files
        print("\n📁 Files in Labels archive:")
        for file in labels_zip.namelist():
            print(f"  • {file}")

        # Read the ML-centric CSV with biomarkers
        print("\n" + "="*80)
        print("📊 BIOMARKER_CLINICAL_DATA (9,408 images with full annotations)")
        print("="*80)

        csv_files = [f for f in labels_zip.namelist() if 'Biomarker_Clinical_Data' in f and f.endswith('.csv')]

        if csv_files:
            print(f"\nReading: {csv_files[0]}")
            with labels_zip.open(csv_files[0]) as f:
                df_biomarker = pd.read_csv(f)

                print(f"\n📊 Dataset Shape: {df_biomarker.shape[0]:,} rows × {df_biomarker.shape[1]} columns")

                print("\n📋 ALL CLINICAL COLUMNS:")
                for i, col in enumerate(df_biomarker.columns, 1):
                    print(f"  {i:2d}. {col}")

                print("\n" + "="*80)
                print("DETAILED COLUMN INFORMATION")
                print("="*80)

                # Show data types and sample values
                print("\n1️⃣ CLINICAL LABELS (Basic Patient Info):")
                clinical_cols = ['Eye', 'BCVA', 'CST', 'patient_id']
                for col in clinical_cols:
                    if col in df_biomarker.columns:
                        dtype = df_biomarker[col].dtype
                        non_null = df_biomarker[col].notna().sum()
                        unique = df_biomarker[col].nunique()
                        sample_values = df_biomarker[col].dropna().unique()[:5]

                        print(f"\n  📌 {col}:")
                        print(f"      Type: {dtype}")
                        print(f"      Non-null values: {non_null:,}/{len(df_biomarker):,}")
                        print(f"      Unique values: {unique:,}")
                        print(f"      Sample values: {sample_values}")

                print("\n2️⃣ BIOMARKERS (Disease Indicators):")
                biomarker_cols = [col for col in df_biomarker.columns if 'fluid' in col.lower() or 'drt' in col.lower() or 'ped' in col.lower() or 'lesion' in col.lower()]

                if not biomarker_cols:
                    # Try to identify biomarker columns differently
                    biomarker_cols = [col for col in df_biomarker.columns if col not in ['Eye', 'BCVA', 'CST', 'patient_id', 'image_id', 'filename']]

                for col in biomarker_cols[:10]:  # Show first 10 biomarkers
                    unique_vals = df_biomarker[col].unique()
                    value_counts = df_biomarker[col].value_counts()

                    print(f"\n  🔬 {col}:")
                    print(f"      Unique values: {unique_vals}")
                    print(f"      Distribution:\n{value_counts}")

                print("\n3️⃣ SAMPLE DATA (First 5 rows):")
                print(df_biomarker.head())

                print("\n4️⃣ STATISTICAL SUMMARY:")
                print(df_biomarker.describe())

        # Read the Excel file with all clinical data
        print("\n" + "="*80)
        print("📊 CLINICAL_DATA (78,000+ images)")
        print("="*80)

        excel_files = [f for f in labels_zip.namelist() if 'Clinical_Data' in f and f.endswith('.xlsx')]

        if excel_files:
            print(f"\nReading: {excel_files[0]}")
            with labels_zip.open(excel_files[0]) as f:
                df_clinical = pd.read_excel(f)

                print(f"\n📊 Dataset Shape: {df_clinical.shape[0]:,} rows × {df_clinical.shape[1]} columns")

                print("\n📋 COLUMNS:")
                for i, col in enumerate(df_clinical.columns, 1):
                    print(f"  {i}. {col}")

                print("\n📊 COLUMN DETAILS:")
                for col in df_clinical.columns:
                    dtype = df_clinical[col].dtype
                    non_null = df_clinical[col].notna().sum()
                    unique = df_clinical[col].nunique()

                    print(f"\n  📌 {col}:")
                    print(f"      Type: {dtype}")
                    print(f"      Non-null: {non_null:,}/{len(df_clinical):,} ({100*non_null/len(df_clinical):.1f}%)")
                    print(f"      Unique values: {unique:,}")

                    if dtype in ['int64', 'float64']:
                        print(f"      Range: {df_clinical[col].min()} to {df_clinical[col].max()}")
                        print(f"      Mean: {df_clinical[col].mean():.2f}")
                    else:
                        sample_values = df_clinical[col].dropna().unique()[:5]
                        print(f"      Sample values: {sample_values}")

                print("\n5️⃣ SAMPLE DATA (First 10 rows):")
                print(df_clinical.head(10))

print("\n" + "="*80)
print("ANALYSIS COMPLETE")
print("="*80)
```
