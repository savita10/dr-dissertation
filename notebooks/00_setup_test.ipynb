```python
# ── CELL 2 | Ensure correct PyTorch for the GPU ──
import subprocess, sys
result = subprocess.run(
    [sys.executable, "-c", "import torch; print(torch.version.cuda)"],
    capture_output=True, text=True
)
if "12.8" not in result.stdout:
    print("Installing PyTorch for this GPU...")
    subprocess.run([
        sys.executable, "-m", "pip", "install",
        "torch==2.7.0", "torchvision==0.22.0",
        "--index-url", "https://download.pytorch.org/whl/cu128",
        "--force-reinstall", "-q"
    ])
    print("PyTorch installed — please restart the runtime, then continue from Cell 2.")
else:
    print(f"PyTorch CUDA {result.stdout.strip()} — correct version already loaded.")
```
